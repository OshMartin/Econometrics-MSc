### Least Squares Regression

You can perform ordinary least squares (OLS) regression using variables in a dataframe. The following command regresses the budget share of food on the logarithms of total expenditure and income using `data`.

````r
lmobj <- lm(wfood ~ log(totexp) + log(income), data = data)

typeof(lmobj)
class(lmobj)

summary(lmobj) # Summary of the fitted model
````

Strictly speaking, `wfood ~ log(totexp) + log(income)` is a formula object. For more details, see [Lam (2010)](https://mirror.nju.edu.cn/CRAN/doc/contrib/Lam-IntroductionToR_LHL.pdf), pp. 142-144. The type of `lmobj` is a list, so list operations can be applied to it. Its class is `lm`. The functions below are useful for this class.
```R
formula(lmobj) #Extract the model formula
coef(lmobj)    # Estimated model parameters (beta-hat OLS)
resid(lmobj)   # Estimated residuals (M_X y)
fitted(lmobj)  # Fitted values (P_X y)
deviance(lmobj) # Residual sum of squares (SSR, or y' M_X y)
confint(lmobj) # Confidence intervals
logLik(lmobj)  # Log-likelihood value, assuming normal errors
AIC(lmobj, k = 2) # k = 2 for AIC; k = log(N) for BIC
vcov(lmobj)   # Estimated variance-covariance matrix of beta-hat OLS
anova(lmobj)  # ANOVA table with explained sum of squares and residual sum of squares
predict(lmobj, data.frame(totexp = 1, income = 1)) # Prediction
plot(lmobj)   # Diagnostic plots
```

To use this function, first install the package. Installation is required only once for a given package, though everytime you want to use the package you have to load it first with the `library()` function below.

````r
# 1. Install the package if it is not already available
if (!requireNamespace("tseries", quietly = TRUE)) install.packages("tseries")
#load the package
library(tseries)

# 2. Generate or load your data (e.g., 100 random normal values)
set.seed(123)
data_normal <- rnorm(100)

# 3. Run the Jarque-Bera test
jb_result <- jarque.bera.test(data_normal)

# 4. View results
print(jb_result)
````

#### Writing Functions

An R function is a procedure that takes some or no inputs and returns outputs. In addition to using built-in functions, you can write your own. The following code creates the Rosenbrock banana function, named `fr`, with `x` as the variable and `a` and `b` as parameters whose default values are 100 and 1, respectively. Here `x` is a $2 \times 1$ vector. In other words,

$$
fr(x_1, x_2) = a(x_2 - x_1^2)^2 + (b - x_1)^2.
$$


````r
fr <- function(x, a = 100, b = 1) {
  # Rosenbrock banana function
  x1 <- x[1]
  x2 <- x[2]
  return(a * (x2 - x1 * x1)^2 + (b - x1)^2)
}

fr(c(0, 0))
````

R's built-in `optim()` function can minimize or maximize a function. The command below minimizes `fr` using $a=101$ and $b=2$ (different from the defaults) and returns the results as a list called `frres`. Adding `control = list(fnscale = -1)` to `optim()` changes minimization to maximization.

````r
frres <- optim(c(0, 0), fr, a = 101, b = 2, hessian = TRUE)
frres

# For maximization, add:
# control = list(fnscale = -1)
````

#### Control the Flow of Execution
When we write a function, we may want it to return different outputs based on the input. The following function uses `if`, `else if`, and `else` to classify a number.

````r
fif <- function(x) {
  if ((x %% 2) == 0) {
    y <- "you have entered an even number."
  } else if ((x %% 2) == 1) {
    y <- "you have entered an odd number."
  } else {
    y <- "what you entered is not an integer."
  }
  return(y)
}
fif(101)
````

Sometimes we want to do something a certain number of times in a function; this is called looping. When we use `for`, we know exactly how many iterations we want to perform.


````r
ffor <- function(x) {
  if (((x %% 2) != 0 && (x %% 2) != 1) || x < 1) {
    stop("Please enter an integer greater than 0.")
  }

  for (i1 in 1:x) {
    print(paste("hi", as.character(i1)))
  }
}
ffor(3)
````

We can use a `while` loop when we are not sure about the number of iterations.

````r
fwhile <- function(x) {
  if (x < 10) {
    stop("Please enter a number greater than or equal to 10.")
  }

  y <- 0
  while (x > 0.1) {
    x <- x / 2
    y <- y + 1
  }
  return(y)
}
fwhile(20)
````
